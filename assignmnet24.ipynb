```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

classification_dataset = pd.read_csv(r"/content/diabetes.csv")
regresssion_dataset = pd.read_csv(r"/content/Chocolate_Sales.csv")
```

```python
# cleaning the classsification dataset
classification_dataset.isnull().sum()
```

```python
classification_dataset = classification_dataset.drop_duplicates()
```

```python
classification_dataset.dtypes
```

```python

classification_dataset.describe()
```

```python
mean_boxes = regresssion_dataset.loc[regresssion_dataset['Boxes_Shipped']>0,'Boxes_Shipped'].mean()
regresssion_dataset.loc[regresssion_dataset['Boxes_Shipped']<0,'Boxes_Shipped'] = int(mean_boxes)
```

```python
regresssion_dataset['Amount'] = pd.to_numeric(regresssion_dataset['Amount'],errors='coerce')
```

```python
amount_mean = regresssion_dataset.loc[regresssion_dataset['Amount'].notna(),'Amount'].mean()
regresssion_dataset['Amount'] = regresssion_dataset['Amount'].fillna(amount_mean)
```

```python
# filling the null values in each columns
for i in classification_dataset.columns:
    if classification_dataset[i].isnull().sum()>0:
        if classification_dataset[i].dtype !='object':
            mean  = classification_dataset.loc[classification_dataset[i].notna(),i].mean()
            classification_dataset[i] = classification_dataset[i].fillna(mean)
        else:
            most_repeted = classification_dataset[i].value_counts().idxmax()
            classification_dataset[i] = classification_dataset[i].fillna(most_repeted)
```

```python
# cheaking the dataset after cleanin
classification_dataset.describe()
```

```python
classification_dataset.isnull().sum()
```

```python
# cleaning the regration dataset
regresssion_dataset.describe()
```

```python
# Cleanin the incorrect values
columns_to_fix = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for col in columns_to_fix:
    mean_val = classification_dataset.loc[classification_dataset[col] > 0, col].mean()
    # Replace negative values with the mean
    classification_dataset.loc[classification_dataset[col] <= 0, col] = int(mean_val)
```

```python
regresssion_dataset = regresssion_dataset.drop_duplicates()
```

```python
regresssion_dataset.describe()
```

```python
regresssion_dataset.isnull().sum()
```

```python
# filling the null values in each columns
for i in regresssion_dataset.columns:
    if regresssion_dataset[i].isnull().sum()>0:
        if regresssion_dataset[i].dtype =='number':
            mean  = regresssion_dataset.loc[regresssion_dataset[i].notna(),i].mean()
            regresssion_dataset[i] = regresssion_dataset[i].fillna(mean)
        else:
            most_repeted = regresssion_dataset[i].value_counts().idxmax()
            regresssion_dataset[i] = regresssion_dataset[i].fillna(most_repeted)
```

```python
classification_dataset.columns
```

```python
# spling the indipendent and targated features of each dataset
x_classification = classification_dataset[['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI','Age']]
y_classfication = classification_dataset['Outcome']
```

```python
x_classification.select_dtypes(include='number').columns
```

```python
regresssion_dataset.columns
```

```python
x_regrassion = regresssion_dataset[['Product', 'Country', 'Channel',
       'Discount_Pct', 'Price_per_Box', 'Marketing_Spend',
       'Boxes_Shipped']]
y_regrassion = regresssion_dataset['Amount']
```

```python
# converting all numerical columns into integer
xc_numcols = x_classification.select_dtypes(include='number').columns
x_classification[xc_numcols] = x_classification[xc_numcols].astype(int)

xr_numcols = x_regrassion.select_dtypes(include='number').columns
x_regrassion[xr_numcols] = x_regrassion[xr_numcols].astype(int)
```

```python
# performing one hot encoding on categorical columns
xc_catcols = x_classification.select_dtypes(include='object').columns
x_classification = pd.get_dummies(x_classification,columns=xc_catcols)


```

```python
xr_catcols = x_regrassion.select_dtypes(include='object').columns
x_regrassion = pd.get_dummies(x_regrassion,columns=xr_catcols)
```

```python
# standard scaling the numerical columns
from sklearn.preprocessing import StandardScaler
xc_scaler = StandardScaler()
x_classification[xc_numcols] = xc_scaler.fit_transform(x_classification[xc_numcols])
xr_scaler = StandardScaler()
x_regrassion[xr_numcols] = xr_scaler.fit_transform(x_regrassion[xr_numcols])
```

```python
# spliting dataset int traning and testing dataframes
from sklearn.model_selection import train_test_split

xc_train,xc_test,yc_train,yc_test = train_test_split(x_classification,y_classfication,random_state=42,shuffle=True,test_size=0.33)
```

```python
xr_train,xr_test,yr_train,yr_test = train_test_split(x_regrassion,y_regrassion,random_state=42,shuffle=True,test_size=0.33)
```

```python
"""Q2. (Classification Algorithms)
On the classification dataset, train and evaluate the following models:
Logistic Regression
Decision Tree Classifier
Support Vector Machine (SVM)
K-Nearest Neighbors (KNN)
Naive Bayes
For each model:
Train the model
Make predictions
Calculate Accuracy, Confusion Matrix, and Classification Report
Create a comparison table of all models’ performance."""

from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression()
logistic_model.fit(xc_train,yc_train)
```

```python
from sklearn.tree import DecisionTreeClassifier

decision_model = DecisionTreeClassifier()
decision_model.fit(xc_train,yc_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier

K_model = KNeighborsClassifier(11)
K_model.fit(xc_train,yc_train)
```

```python
from sklearn.naive_bayes import GaussianNB
Navi_model = GaussianNB()
Navi_model.fit(xc_train,yc_train)
```

```python
models = {
    'logistic_model':logistic_model,
    'decision-tree_model':decision_model,
    'KNN_model':K_model,
    'Navi-bias_model':Navi_model

}
```

```python
# confusion matrix for each model
from sklearn.metrics import confusion_matrix
for i in models:
    x = models[i]
    y_pred = x.predict(xc_test)
    print(f"Confusion matrix of:{i}\n{confusion_matrix(yc_test,y_pred)}")
```

```python
# Accuracy of each model
from sklearn.metrics import accuracy_score
for i in models:
    x = models[i]
    y_pred = x.predict(xc_test)
    print(f"Accuracy for:{i}\n{accuracy_score(yc_test,y_pred)}")
```

```python
# classification report for each Model
from sklearn.metrics import classification_report
for i in models:
    x = models[i]
    y_pred = x.predict(xc_test)
    print(f"Classification Report for:{i}\n{classification_report(yc_test,y_pred)}")
```

```python
reports = {}

for name, model in models.items():
    y_pred = model.predict(xc_test)
    report = classification_report(yc_test, y_pred, output_dict=True)
    reports[name] = report['weighted avg']  # use weighted avg for overall comparison

# Convert to DataFrame
comparison_table = pd.DataFrame(reports).T
print(comparison_table)
```
